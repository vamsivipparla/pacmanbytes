```python
from sys import stdout
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math
import statistics
import scipy.stats   # for 2-sample Kolmogorov-Smirnov test
#from mlxtend.evaluate import bias_variance_decomp
```

read data

```python
df =pd.read_excel('results.xlsx','result') #get reference value_
data =np.array(df)
#print(data)
MM0 = np.array(data[0:8,2:5],dtype=float)
MM = np.array(data[8:16,2:5],dtype=float)
DFS = np.array(data[16:24,2:5],dtype=float)
BFS = np.array(data[24:32,2:5],dtype=float)
UCS = np.array(data[32:40,2:5],dtype=float)
Astar = np.array(data[40:48,2:5],dtype=float)
print(MM)
```

T-test

```python
import scipy.stats
# Perform the unpaired Welch’s t-test on Node expanded against MM
print('Nodes expanded')
sta, result = scipy.stats.ttest_ind(MM[:,0], MM0[:,0], equal_var=False)#, nan_policy = 'omit', alternative='less')
print ('MM0: ',result)
sta, result = scipy.stats.ttest_ind(MM[:,0], DFS[:,0], equal_var=False)#, nan_policy = 'omit', alternative='greater')
print ('DFS: ',result)
sta, result = scipy.stats.ttest_ind(MM[:,0], BFS[:,0], equal_var=False)#, nan_policy = 'omit', alternative='less')
print ('BFS: ',result)
sta, result = scipy.stats.ttest_ind(MM[:,0], UCS[:,0], equal_var=False)#, nan_policy = 'omit', alternative='less')
print ('UCS: ',result)
sta, result = scipy.stats.ttest_ind(MM[:,0], Astar[:,0], equal_var=False)#, nan_policy = 'omit', alternative='less')
print ('A*: ',result)

# Perform the unpaired Welch’s t-test on Path Cost against MM
print('Path cost')
sta, result = scipy.stats.ttest_ind(MM[:,1], MM0[:,1], equal_var=False)#, nan_policy = 'omit', alternative='greater')
print ('MM0: ',result)
sta, result = scipy.stats.ttest_ind(MM[:,1], DFS[:,1], equal_var=False)#, nan_policy = 'omit', alternative='less')
print ('DFS: ',result)
sta, result = scipy.stats.ttest_ind(MM[:,1], BFS[:,1], equal_var=False)#, nan_policy = 'omit', alternative='greater')
print ('BFS: ',result)
sta, result = scipy.stats.ttest_ind(MM[:,1], UCS[:,1], equal_var=False)#, nan_policy = 'omit', alternative='greater')
print ('UCS: ',result)
sta, result = scipy.stats.ttest_ind(MM[:,1], Astar[:,1], equal_var=False)#, nan_policy = 'omit', alternative='greater')
print ('A*: ',result)

# Perform the unpaired Welch’s t-test on Score against MM
print('Score')
sta, result = scipy.stats.ttest_ind(MM[:,2], MM0[:,2], equal_var=False)#, nan_policy = 'omit', alternative='greater')
print ('MM0: ',result)
sta, result = scipy.stats.ttest_ind(MM[:,2], DFS[:,2], equal_var=False)#, nan_policy = 'omit', alternative='less')
print ('DFS: ',result)
sta, result = scipy.stats.ttest_ind(MM[:,2], BFS[:,2], equal_var=False)#, nan_policy = 'omit', alternative='greater')
print ('BFS: ',result)
sta, result = scipy.stats.ttest_ind(MM[:,2], UCS[:,2], equal_var=False)#, nan_policy = 'omit', alternative='greater')
print ('UCS: ',result)
sta, result = scipy.stats.ttest_ind(MM[:,2], Astar[:,2], equal_var=False)#, nan_policy = 'omit', alternative='greater')
print ('A*: ',result)
```

PCA

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sn

pca=PCA(n_components=3)
data_pca=pca.fit_transform(data[0:48,2:5])
print(data_pca.shape)
data_pca=pca.fit_transform(data[0:48,2:5])[:,(0,1)]
print(data_pca.shape)
labels1 = data[0:48,0]
data_pca=np.column_stack((data_pca,labels1))
print(data_pca.shape)
pca_df=pd.DataFrame(data=data_pca, columns=("PC-1","PC-2","labels"))
sn.FacetGrid(pca_df, hue="labels", size=5).map(plt.scatter,'PC-1','PC-2').add_legend()
plt.show()

data_pca=pca.fit_transform(data[0:48,2:5])[:,(1,2)]
print(data_pca.shape)
labels1 = data[0:48,0]
data_pca=np.column_stack((data_pca,labels1))
print(data_pca.shape)
pca_df=pd.DataFrame(data=data_pca, columns=("PC-2","PC-3","labels"))
sn.FacetGrid(pca_df, hue="labels", size=5).map(plt.scatter,'PC-2','PC-3').add_legend()
plt.show()
```
