```python
from sys import stdout
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math
import statistics
import scipy.stats   # for 2-sample Kolmogorov-Smirnov test
#from mlxtend.evaluate import bias_variance_decomp
```

read data

```python
df =pd.read_excel('results.xlsx','New_result') #get reference value_
data =np.array(df)
Astar = np.array(data[0:10,2:5],dtype=float)
BFS = np.array(data[10:20,2:5],dtype=float)
DFS = np.array(data[20:30,2:5],dtype=float)
MM_diag = np.array(data[30:40,2:5],dtype=float)
MM_euc = np.array(data[40:50,2:5],dtype=float)
MM_man = np.array(data[50:60,2:5],dtype=float)
MM0 = np.array(data[60:70,2:5],dtype=float)
UCS = np.array(data[70:80,2:5],dtype=float)
print(MM_euc)
```

T-test

scipy.stats.ttest_ind() for unpaired

scipy.stats.ttest_rel() for paired

```python
import scipy.stats
# Perform the paired Welch’s t-test on Node expanded against MM_diag
print('Nodes expanded')
sta, result = scipy.stats.ttest_rel(MM_diag[:,0], MM_euc[:,0])
print ('MM_euc: ',result)
sta, result = scipy.stats.ttest_rel(MM_diag[:,0], MM_man[:,0],alternative='greater')
print ('MM_man: ',result)
sta, result = scipy.stats.ttest_rel(MM_diag[:,0], MM0[:,0],alternative='less')
print ('MM0: ',result)
sta, result = scipy.stats.ttest_rel(MM_diag[:,0], DFS[:,0],alternative='less')
print ('DFS: ',result)
sta, result = scipy.stats.ttest_rel(MM_diag[:,0], BFS[:,0],alternative='less')
print ('BFS: ',result)
sta, result = scipy.stats.ttest_rel(MM_diag[:,0], UCS[:,0],alternative='less')
print ('UCS: ',result)
sta, result = scipy.stats.ttest_rel(MM_diag[:,0], Astar[:,0],alternative='less')
print ('A*: ',result)
print('Path cost')
# Perform the paired Welch’s t-test on Path Cost against MM
sta, result = scipy.stats.ttest_rel(MM_diag[:,1], MM_euc[:,1],alternative='less')
print ('MM_euc: ',result)
sta, result = scipy.stats.ttest_rel(MM_diag[:,1], MM_man[:,1],alternative='less')
print ('MM_man: ',result)
sta, result = scipy.stats.ttest_rel(MM_diag[:,1], MM0[:,1],alternative='less')
print ('MM0: ',result)
sta, result = scipy.stats.ttest_rel(MM_diag[:,1], DFS[:,1],alternative='less')
print ('DFS: ',result)
sta, result = scipy.stats.ttest_rel(MM_diag[:,1], BFS[:,1],alternative='less')
print ('BFS: ',result)
sta, result = scipy.stats.ttest_rel(MM_diag[:,1], UCS[:,1],alternative='less')
print ('UCS: ',result)
sta, result = scipy.stats.ttest_rel(MM_diag[:,1], Astar[:,1],alternative='less')
print ('A*: ',result)
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sn

pca=PCA(n_components=2)
data_pca=pca.fit_transform(data[0:80,2:4])
print(data_pca.shape)
data_pca=pca.fit_transform(data[0:80,2:4])[:,(0,1)]
print(data_pca.shape)
labels1 = data[0:80,0]
data_pca=np.column_stack((data_pca,labels1))
print(data_pca.shape)
pca_df=pd.DataFrame(data=data_pca, columns=("PC-1","PC-2","labels"))
sn.FacetGrid(pca_df, hue="labels", size=5).map(plt.scatter,'PC-1','PC-2').add_legend()
plt.show()
```
